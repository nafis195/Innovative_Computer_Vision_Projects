# Class 4 - Edge Detection and Image Segmentation

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# loading image
img = cv2.imread("/content/taj.png")
plt.imshow(img)
plt.show()
```

```python
# sobel operator
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3) #Gx
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3) #Gy

magnitude = np.sqrt(sobelx**2 + sobely**2)
magnitude = np.uint8(np.clip(magnitude, 0, 255))
magnitude_display = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX)

plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1), plt.title('Original'), plt.imshow(img, cmap="gray"), plt.axis("off")
plt.subplot(1, 3, 2), plt.title('Sobel Magnitude'), plt.imshow(magnitude_display, cmap="gray"), plt.axis("off")
plt.subplot(1, 3, 3), plt.title('Sobel X + Y'), plt.imshow(cv2.addWeighted(cv2.convertScaleAbs(sobelx), 0.5, cv2.convertScaleAbs(sobely), 0.5, 0)), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# prewitt operator
kernelx = np.array([[-1, 0, 1],
                    [-1, 0, 1],
                    [-1, 0, 1]], dtype=np.float32)

kernely = np.array([[-1, -1, -1],
                    [0, 0, 0],
                    [1, 1, 1]], dtype=np.float32)

prewittx = cv2.filter2D(img, -1, kernelx)
prewitty = cv2.filter2D(img, -1, kernely)
prewitt_combined = cv2.addWeighted(prewittx, 0.5, prewitty, 0.5, 0)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(prewittx, cmap='gray'), plt.title('Prewitt X'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(prewitty, cmap='gray'), plt.title('Prewitt Y'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(prewitt_combined, cmap='gray'), plt.title('Prewitt Combined'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# scharr operator
scharr_x = cv2.Scharr(img, cv2.CV_64F, 1, 0)
scharr_y = cv2.Scharr(img, cv2.CV_64F, 0, 1)
scharr_magnitude = cv2.magnitude(scharr_x, scharr_y)

scharr_x = np.uint8(np.absolute(scharr_x))
scharr_y = np.uint8(np.absolute(scharr_y))
scharr_magnitude = np.uint8(np.clip(scharr_magnitude, 0, 255))

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(scharr_x, cmap='gray'), plt.title('Scharr X'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(scharr_y, cmap='gray'), plt.title('Scharr Y'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(scharr_magnitude, cmap='gray'), plt.title('Scharr Magnitude'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# canny edge detector

blurred = cv2.GaussianBlur(img, (5, 5), 1.4)
edges = cv2.Canny(blurred, threshold1=50, threshold2=150)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1), plt.imshow(img, cmap='gray'), plt.title('Original Image'), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(edges, cmap='gray'), plt.title('Canny Edges'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# laplacian of gaussian

blurred = cv2.GaussianBlur(img, (3, 3), 0)
log = cv2.Laplacian(blurred, cv2.CV_64F)
log_abs = cv2.convertScaleAbs(log)

plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1), plt.imshow(img, cmap="gray"), plt.title("Original"), plt.axis("off")
plt.subplot(1, 3, 2), plt.imshow(blurred, cmap="gray"), plt.title("Blurred"), plt.axis("off")
plt.subplot(1, 3, 3), plt.imshow(log_abs, cmap="gray"), plt.title("Laplacian of Gaussian"), plt.axis("off")
plt.tight_layout()
plt.show()
```

# Class 6 - Image Segmentation

```python
# import libraries
import cv2
import matplotlib.pyplot as plt
import numpy as np
# %matplotlib widget
```

```python
# load and visualize image
img = cv2.imread("/content/cell.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# image shape
img.shape
```

```python
# convert to grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# display grayscale
plt.figure(figsize=(6, 6))
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# seed selection
def onclick(event):
  global seed_point
  if event.xdata is not None and event.ydata is not None:
    seed_point = (int(event.xdata), int(event.ydata))
    print("Selected seed point", seed_point)

    segmented = region_growing(img_gray, seed_point, threshold)
    plt.figure(figsize=(6, 6))
    plt.imshow(segmented, cmap="gray")
    plt.title(f"Region Grown from Seed {seed_point}")
    plt.axis("off")
    plt.show()

fig, ax = plt.subplots(figsize=(6, 6))
ax.imshow(img_gray, cmap="gray")
ax.set_title("Select Seed Point")
cid = fig.canvas.mpl_connect("button_press_event", onclick)
plt.show()
```

```python
# set threshold and seed point
threshold = 80
seed_point = (100, 100)
```

```python
# region growing function
def region_growing(image, seed, threshold):
  H, W = image.shape
  segmented = np.zeros((H, W), np.uint8)
  visited = np.seros_like(image, dtype="bool")
  seed_value = int(image[seed])

  to_process = [seed]

  while to_process:
    y, x = to_process.pop()
    if visited[y, x]:
      continue
    visited[y, x] = True

    pixel_value = int(image[y, x])
    if abs(pixel_value - seed_value) <= threshold:
      segmented[y, x] = 255

      # 4 neighbors
      if x > 0:
        to_process.append((y, x - 1))
      if x < W - 1:
        to_process.append((y, x + 1))
      if y > 0:
        to_process.append((y - 1, x))
      if y < H - 1:
        to_process.append((y + 1, x))

  return segmented
```

```python
# apply region growing and showing image
segmented = region_growing(img_gray, seed_point, threshold)

plt.figure(figsize=(6, 6))
plt.imshow(segmented, cmap="gray")
plt.title(f"Region Grown from Seed {seed_point}")
plt.axis("off")
plt.show()
```

# Class 7 - Image Transformation

```python
# importing libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("/content/alzheimer.png")
rows, cols = img.shape[:2]
```

```python
plt.imshow(img)
plt.show()
```

**Affine Transformation**

```python
# translation - move the image in x and y direction
M_translated = np.float32([[1, 0, 50], [0, 1,30]]) # move 50 right, 30 down
translated = cv2.warpAffine(img, M_translated, (cols, rows))
```

```python
# scaling - resizes the image using a scaling factor
scaled = cv2.resize(img, None, fx=1.2, fy=0.7) # scale 120% in width, 70% in height
```

```python
# rotation - rotates image around its center by a given angle and optional scale
center = (cols/2, rows/2)
M_rotate = cv2.getRotationMatrix2D(center, angle=45, scale=1) # rotate 45 degrees
rotated = cv2.warpAffine(img, M_rotate, (cols, rows))
```

```python
# shearing - distorts the shape along x or y axis
M_shear = np.float32([[1, 0.5, 0], [0.2, 1, 0]])
sheared = cv2.warpAffine(img, M_shear, (int(cols*1.5), int(rows*1.5)))
```

```python
# combination - multiple transformation and combine effects
M1 = cv2.getRotationMatrix2D(center, angle=30, scale=0.8) # rotation + scaling
M1[0, 2] += 40 # add translation
M1[1, 2] += 30 # add translation y
combined = cv2.warpAffine(img, M1, (cols, rows))
```

```python
# plotting and showing images
plt.figure(figsize=(12,8))
plt.subplot(231), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title("Original")
plt.subplot(232), plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB)), plt.title("Translated")
plt.subplot(233), plt.imshow(cv2.cvtColor(scaled, cv2.COLOR_BGR2RGB)), plt.title("Scaled")
plt.subplot(234), plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)), plt.title("Rotated")
plt.subplot(235), plt.imshow(cv2.cvtColor(sheared, cv2.COLOR_BGR2RGB)), plt.title("Sheared")
plt.subplot(236), plt.imshow(cv2.cvtColor(combined, cv2.COLOR_BGR2RGB)), plt.title("Combined")
plt.tight_layout()
plt.show()
```

```python
# define 4 points in the source image
pts_src = np.float32([[30, 30], [30, 70], [150, 30], [150, 170]])
```

```python
# define destination points to get a straight rectangle
pts_dst = np.float32([[0, 0], [0, 200], [200, 0], [200, 200]])
```

```python
# compute the perspective transform matrix
H = cv2.getPerspectiveTransform(pts_src, pts_dst)

# apply wrap
wraped_img = cv2.warpPerspective(img, H, (250, 250))
```

```python
# show images
plt.figure(figsize=(10, 5))
plt.subplot(121), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title("Original")
plt.subplot(122), plt.imshow(cv2.cvtColor(wraped_img, cv2.COLOR_BGR2RGB)), plt.title("Prespective Corrected")
plt.show()
```

# Class 8 - Image Morphology

```python
# importing libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# showing original image
def show_img(title, img):
  plt.imshow(img, cmap="gray")
  plt.title(title)
  plt.axis('off')
  plt.show()

img = cv2.imread('/content/fingerprint.png', 0)
```

```python
show_img("Original Image", img)
```

```python
# erosion image
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
kernel = np.ones((3, 3), np.uint8)
erosion = cv2.erode(binary, kernel, iterations=1)
```

```python
show_img("Eroded Image", erosion)
```

```python
# dilated image
dilated = cv2.dilate(binary, kernel, iterations=1)
show_img("Dilated Image", dilated)
```

```python
# image opening
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
show_img("Image Opening", opening)
```

```python
# image closing
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)
show_img("Image Closing", closing)
```

```python
operations = {
    'Original': binary,
    'Erosion': cv2.erode(binary, kernel),
    'Dilation': cv2.dilate(binary, kernel),
    'Opening': cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel),
    'Closing': cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel),
}

plt.figure(figsize=(15, 5))
for i, (name, img) in enumerate(operations.items(), 1):
    plt.subplot(1, 5, i)
    plt.imshow(img, cmap='gray')
    plt.title(name)
    plt.axis('off')
plt.show()
```

```python
# image gradient
gradient = cv2. morphologyEx(binary, cv2.MORPH_GRADIENT, kernel)
show_img("Image Gradient", gradient)
```

```python
# tophat transformation
tophat = cv2.morphologyEx(binary, cv2.MORPH_TOPHAT, kernel)
show_img("Tophat Transformation", tophat)
```

```python
# blackhat transformation
blackhat = cv2.morphologyEx(binary, cv2.MORPH_BLACKHAT, kernel)
show_img("Blackhat Transformation", blackhat)
```

```python
advanced_ops = {
    'Original': binary,
    'Gradient': cv2.morphologyEx(binary, cv2.MORPH_GRADIENT, kernel),
    'Top-hat': cv2.morphologyEx(binary, cv2.MORPH_TOPHAT, kernel),
    'Black-hat': cv2.morphologyEx(binary, cv2.MORPH_BLACKHAT, kernel),
}

plt.figure(figsize=(15, 5))
for i, (name, img) in enumerate(advanced_ops.items(), 1):
    plt.subplot(1, 4, i)
    plt.imshow(img, cmap='gray')
    plt.title(name)
    plt.axis('off')
plt.show()
```
